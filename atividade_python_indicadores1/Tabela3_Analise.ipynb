```python
bruto = pd.read_excel(path, sheet_name="Sheet1", header=None)
```

```python
bruto
```

```python
regioes = bruto.iloc[3:9].copy()

regioes.columns = [
    "regiao",
    "superior_total", "superior_homens", "superior_mulheres",
    "ctem_total", "ctem_homens", "ctem_mulheres",
    "edu_total", "edu_homens", "edu_mulheres",
]

regioes = regioes.reset_index(drop=True)
regioes
```

```python
for col in regioes.columns[1:]:
    regioes[col] = pd.to_numeric(regioes[col], errors="coerce")

regioes
```

```python
curso_superior = regioes[["regiao", "superior_total", "superior_homens", "superior_mulheres"]].copy()
curso_superior.columns = ["regiao", "total", "homens", "mulheres"]
curso_superior
```

```python
ctem = regioes[["regiao", "ctem_total", "ctem_homens", "ctem_mulheres"]].copy()
ctem.columns = ["regiao", "total", "homens", "mulheres"]
ctem
```

```python
edu_saude = regioes[["regiao", "edu_total", "edu_homens", "edu_mulheres"]].copy()
edu_saude.columns = ["regiao", "total", "homens", "mulheres"]
edu_saude
```

```python
print("Curso superior (total):", (curso_superior["mulheres"] > curso_superior["homens"]).sum(), "de", len(curso_superior), "regiões")
print("CTEM:                   ", (ctem["mulheres"] > ctem["homens"]).sum(), "de", len(ctem), "regiões")
print("Educação/Saúde/Bem-estar:", (edu_saude["mulheres"] > edu_saude["homens"]).sum(), "de", len(edu_saude), "regiões")
```
