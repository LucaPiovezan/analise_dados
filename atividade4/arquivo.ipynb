```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\lucot\Downloads\3 etapa\atividade4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx", header=2)
df.drop(columns=["Unnamed: 0"], inplace=True)
df.head()
```

```python
df.columns.tolist()
```

```python
df.rename(columns={
    'UF': 'UF',
    'COD. UF': 'COD_UF',
    'COD. MUNIC': 'COD_MUNIC',
    'NOME DO MUNICÍPIO': 'NOME_MUNICIPIO',
    'População Município 2010\n(Sinopse)': 'POP_2010_SINOSE',
    'População 2010 (Alterações de Limites até 2022)1': 'POP_2010',
    'População Censo 2022': 'POP_2022'
}, inplace=True)
df.head()
```

```python
df = df[df["UF"].str.len() == 2]
pop_estado = df.groupby("UF")[["POP_2010", "POP_2022"]].sum().reset_index()
pop_estado.head()
```

```python
pop_estado["CRESCIMENTO"] = pop_estado["POP_2022"] - pop_estado["POP_2010"]
pop_estado = pop_estado.sort_values("CRESCIMENTO", ascending=False).reset_index(drop=True)
pop_estado
```

```python
pop_estado.to_csv(r"C:\Users\lucot\Downloads\3 etapa\atividade4\populacao_estado.csv", sep=";", index=False)
```

```python
pop_municipio = df.groupby(["UF", "NOME_MUNICIPIO"])[["POP_2010", "POP_2022"]].sum().reset_index()
pop_municipio.head()
```

```python
pop_municipio["CRESCIMENTO"] = pop_municipio["POP_2022"] - pop_municipio["POP_2010"]
pop_municipio = pop_municipio.sort_values("CRESCIMENTO", ascending=False).reset_index(drop=True)
pop_municipio
```

```python
pop_municipio.to_csv(r"C:\Users\lucot\Downloads\3 etapa\atividade4\populacao_municipio.csv", sep=";", index=False)
```
